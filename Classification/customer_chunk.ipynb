```python
# impport libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

import re
import nltk
```

```python
# load data
data = pd.read_csv("D:/D/Machine Learning/data/Classification/Customer Chunk/customer_chunk.csv")
data.head(5)
```

```python
# check the type of the data
data.dtypes
```

```python
# check null values and remove it
data = data.dropna()
data.isnull().sum()
```

```python
# convert object number into int
data['TotalCharges'] = data['TotalCharges'].apply(lambda x: int(float(x)) if str(x).strip().replace('.', '', 1).isdigit() else np.nan)
data.head(1)
```

```python
# processign object data
column_object = data.select_dtypes(include=['object']).columns
column_object
```

```python
stop_word = set(stopwords.words('english'))
def remove_stopword(txt):
    word_token = word_tokenize(str(txt).lower())
    flt = [word for word in word_token if word.isalpha() and word not in stop_word]
    return " ".join(flt)

def cleaning_text(data):
    dum_data = data
    column_object = dum_data.select_dtypes(include=['object']).columns
    
    # Normalizae text
    for i in column_object:
        dum_data[i] = dum_data[i].str.lower()
        dum_data[i] = dum_data[i].str.strip()

    # Remove punctuation
    for i in column_object:
        dum_data[i] = data[i].str.replace(r'[^\w\s]', '', regex=True)

    # Stop word function
    # for i in column_object:
    #     dum_data[i] = dum_data[i].apply(remove_stopword)
    return dum_data

data = cleaning_text(data)
data =data.dropna()
data.head(3)
```

```python
# Label encoding
from sklearn import preprocessing
label = preprocessing.LabelEncoder()

def label_encoding(data):
    dum_data = data
    column = dum_data.select_dtypes(include=['object']).columns
    encoders = {}
    for i in column:
        label = preprocessing.LabelEncoder()
        dum_data[i] = label.fit_transform(dum_data[i])
        encoders[i] = label

    return dum_data

data = label_encoding(data)
data.head(3)
```

```python
# try to find relation between data
import seaborn as sns
import matplotlib.pyplot as plt

corr = data.corr()
plt.figure(figsize=(15,15))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.9)
plt.title("Correlation Heatmap")
plt.show()
```
